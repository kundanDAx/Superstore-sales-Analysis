```python
# Total sum of sales
df["Sales"].sum()
```

```python
# Total avg of sales
df["Sales"].mean()
```

```python
# Catergory wise sales
Category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending = False)
```

```python
Category_sales
```

```python
Region_sales = df.groupby("Region")["Sales"].sum().sort_values(ascending = False)
```

```python
Region_sales
```

```python
df["Month"] = df["Order Date"].dt.month_name()
```

```python
Month_sales = df.groupby("Month")["Sales"].sum()
```

```python
Month_sales
```

```python
df["Year"] = df["Order Date"].dt.year
```

```python
Year_sales = df.groupby("Year")["Sales"].sum()
```

```python
Year_sales
```

```python
subcategory_sales = df.groupby("Sub-Category")["Sales"].sum().sort_values(ascending = False)
```

```python
subcategory_sales
```

```python
Top_10_sales = df.groupby("Product ID")["Sales"].sum().sort_values(ascending = False).head(10)
```

```python
Top_10_sales
```

```python
segment_sales = df.groupby("Segment")["Sales"].sum().sort_values(ascending=False)
```

```python
segment_sales
```

```python
import matplotlib.pyplot as plt
```

```python
Top_10_sales.sort_values().plot(kind="barh", figsize=(8,5))
plt.title("Top 10 Product By Sales")
plt.xlabel("Sales")
plt.ylabel("Product Name")
plt.show()
```

```python
Month_sales.plot(kind="barh",
                 figsize=(8,5))
plt.title("Month wise Sales")
plt.xlabel("Sales")
plt.ylabel("Month")
plt.show()
```

```python
Year_sales.plot(kind="pie",figsize=(6,6), autopct="%1.1f%%")
plt.title("Year wise Sales")
plt.ylabel("")
plt.show()
```

```python
Region_sales.plot(kind="barh",figsize=(8,5))
plt.title("Region wise Sales")
plt.xlabel("Sales")
plt.ylabel("Region")
plt.show()
```

```python
subcategory_sales.plot(kind="barh",figsize=(8,5))
plt.title("Subcategory wise Sales")
plt.xlabel("Sales")
plt.ylabel("Subcategory")
plt.show()
```

```python
Category_sales.plot(kind="barh",figsize=(8,5))
plt.title("category wise Sales")
plt.xlabel("Sales")
plt.ylabel("category")
plt.show()
```

```python
(segment_sales/100000).plot(kind="bar",figsize=(8,5))
plt.title("segment wise Sales")
plt.xlabel("Segment")
plt.ylabel("sales(Lakh)")
plt.show()
```

```python
print("Highest Category Name  - ",
Category_sales.idxmax())
print("highest Category sales - ",
Category_sales.max())
```
